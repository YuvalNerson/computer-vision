# Collect Assignment 2A

Set the public GitHub link once the assignment files are published. The final cell creates the code archive and the five-notebook VPL archive.


```python
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

GITHUB_URL = "https://github.com/YuvalNerson/computer-vision/tree/main/assignment-2-computer-vision"
assignment = Path("/content/drive/MyDrive") / FOLDERNAME
notebooks = (
    "BatchNormalization.ipynb", "Dropout.ipynb", "ConvolutionalNetworks.ipynb",
    "PyTorch.ipynb", "RNN_Captioning_pytorch.ipynb",
)
missing = [name for name in notebooks if not (assignment / name).is_file()]
if missing:
    raise FileNotFoundError(f"Missing notebooks: {missing}")

package = assignment / "icv83551"
sources = [
    path for path in package.rglob("*")
    if path.is_file()
    and path.suffix in {".py", ".pyx", ".sh", ".png", ".jpg"}
    and not {"__pycache__", "build", "cifar-10-batches-py", "coco_captioning"}.intersection(path.parts)
]
with ZipFile(assignment / "a2_code_submission.zip", "w", ZIP_DEFLATED) as archive:
    for path in [*(assignment / name for name in notebooks), assignment / "collect_submission.ipynb", *sorted(sources)]:
        archive.write(path, path.relative_to(assignment))
with ZipFile(assignment / "a2_vpl_notebooks.zip", "w", ZIP_DEFLATED) as archive:
    for name in notebooks:
        archive.write(assignment / name, name)
print("Created a2_code_submission.zip and a2_vpl_notebooks.zip")
```
